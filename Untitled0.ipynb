# 1. Coordenadas medidas

```python
loc=ee.Geometry.Point([-69.6727495313754, -0.566971691572036]) #Ubicación

imagen = (ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED") \
          .filterBounds(loc)
          .filterDate("2023-01-01", "2024-12-31")
          .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 10))
          )
imagen_limpia = imagen.median() #Valor medio
```

```python
NDWI = imagen_limpia.select('B3').subtract(imagen_limpia.select('B8')).divide(imagen_limpia.select('B3').add(imagen_limpia.select('B8'))).rename('NDWI')

valor_NDWI = NDWI.reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc,
    scale=10
).get('NDWI').getInfo()

print(f"NDWI (B3-B8/B3+B8): {valor_NDWI}")
```

```python
turbidez = imagen_limpia.select('B4').divide(imagen_limpia.select('B3')).rename('B4_B3')
valor_turbidez = turbidez.reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc,
    scale=10
).get('B4_B3').getInfo()

print(f"Turbidez (B4/B3): {valor_turbidez}")
```

```python
bandas = ['B2', 'B3', 'B4', 'B8', 'B11', 'B12']
Valores = imagen_limpia.select(bandas).reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc,
    scale=10
).getInfo()

for banda, valor in Valores.items():
    print(f"Reflectancia real en {banda}: {valor * 0.0001:.3f}")
```
