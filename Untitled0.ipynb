# 1. Visualización general del área de estudio con coordenadas medidas en campo

```python
loc=ee.Geometry.Point([-69.6727495313754, -0.566971691572036]) #Ubicación
```

```python
Map = geemap.Map()
Map.add_basemap('SATELLITE')
Map.centerObject(loc, 14)
Map
```

```python
imagen = (ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED") \
          .filterBounds(loc)
          .filterDate("2023-01-01", "2023-12-31")
          .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 10))
          )
imagen_limpia = imagen.sort("CLOUDY_PIXEL_PERCENTAGE").first()
```

```python
NDWI_1 = imagen_limpia.select('B3').subtract(imagen_limpia.select('B8')).divide(imagen_limpia.select('B3').add(imagen_limpia.select('B8'))).rename('NDWI')

valor_NDWI_1  = NDWI_1 .reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc,
    scale=10
).get('NDWI').getInfo()

print(f"NDWI (B3-B8/B3+B8): {valor_NDWI_1}")
```

```python
turbidez_1  = imagen_limpia.select('B4').divide(imagen_limpia.select('B3')).rename('B4_B3')
valor_turbidez_1  = turbidez_1.reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc,
    scale=10
).get('B4_B3').getInfo()

print(f"Turbidez (B4/B3): {valor_turbidez_1 }")
```

```python
bandas = ['B2', 'B3', 'B4', 'B8', 'B11', 'B12']
Valores_imag_1 = imagen_limpia.select(bandas).reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc,
    scale=10
).getInfo()

for bandas, valor_imag_1 in Valores_imag_1.items():
    print(f"Reflectancia real en {bandas}: {valor_imag_1  * 0.0001:.3f}")
```

```python
print(f"Reflectancia real en B11: {Valores_imag_1['B11'] * 0.0001:.3f}")
```

# Visualización general del área de estudio y coordenadas centrales de la laguna.

```python
viz = {
    'bands': ['B4', 'B3', 'B2'],
    'min': 0,
    'max': 3000,
    'gamma': 1.4

}

ndwi_viz = {
    'min': -0.5,
    'max': 0.5,
    'palette': ['red', 'white', 'blue'] # Rojo=Vegetación, Blanco=Suelo, Azul=Agua
}


Map.addLayer(imagen_limpia, viz, 'Laguna Gredal Limpia')
#Map.addLayer(NDWI, ndwi_viz, 'Capa NDWI')
Map.centerObject(loc, 14)
Map.addLayer(loc, {"color": "yellow"}, "loc")

Map
```

# Punto central del poligono

```python
mi_poligono = Map.draw_last_feature.geometry()
```

```python
centro_real = mi_poligono.centroid(1)
loc_cent_poli = centro_real.coordinates().getInfo()
```

```python
print(f"coordenadas: {loc_cent_poli}")
```

```python
loc_punt_centr = ee.Geometry.Point(loc_cent_poli)

bandas = ['B2', 'B3', 'B4', 'B8', 'B11', 'B12']
Valores_imag_2 = imagen_limpia.select(bandas).reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc_punt_centr,
    scale=10
).getInfo()

for bandas, valor in Valores_imag_2.items():
    print(f"Reflectancia real en {bandas}: {valor * 0.0001:.3f}")
```

```python
NDWI_punt_centr = imagen_limpia.select('B3').subtract(imagen_limpia.select('B8')).divide(imagen_limpia.select('B3').add(imagen_limpia.select('B8'))).rename('NDWI')

valor_NDWI_2 = NDWI_punt_centr.reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc_punt_centr,
    scale=10
).get('NDWI').getInfo()

print(f"NDWI (B3-B8/B3+B8): {valor_NDWI_2}")
```

```python
turbidez_2= imagen_limpia.select('B4').divide(imagen_limpia.select('B3')).rename('B4_B3')
valor_turbidez_2 = turbidez_2.reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc_punt_centr,
    scale=10
).get('B4_B3').getInfo()

print(f"Turbidez (B4/B3): {valor_turbidez_2}")
```

# Ubicación segun el mejor valor de NDWI

```python
area= loc.buffer(100)
ndwi_area = imagen_limpia.normalizedDifference(['B3', 'B8']).rename('NDWI')
muestras = ndwi_area.sample(region=area, scale=10, geometries=True)
mejor_punto = muestras.sort('NDWI', False).first()

loc_ndwi = mejor_punto.geometry().coordinates().getInfo()
valor_NDWI_3 = mejor_punto.get('NDWI').getInfo()

print(f"Valor de NDWI en el punto: {valor_NDWI_3}")
print(loc_ndwi)
```

```python
loc_agua_NDWI = ee.Geometry.Point(loc_ndwi)

bandas = ['B2', 'B3', 'B4', 'B8', 'B11', 'B12']
Valores_imag_3 = imagen_limpia.select(bandas).reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc_agua_NDWI,
    scale=10
).getInfo()

for bandas, valor in Valores_imag_3.items():
    print(f"Reflectancia real en {bandas}: {valor * 0.0001:.3f}")
```

```python
turbidez_3 = imagen_limpia.select('B4').divide(imagen_limpia.select('B3')).rename('B4_B3')
valor_turbidez_3 = turbidez_3.reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc_agua_NDWI,
    scale=10
).get('B4_B3').getInfo()

print(f"Turbidez (B4/B3): {valor_turbidez_3}")
```
