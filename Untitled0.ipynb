```python
loc=ee.Geometry.Point([-69.6728045225969,-0.566948200364969]) #Ubicación

imagen = (ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED") \
          .filterBounds(loc)
          .filterDate("2025-01-01", "2025-12-31")
          .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 10))
          )
imagen_limpia = imagen.median() #Valor medio

viz = {
    'bands': ['B4', 'B3', 'B2'],
    'min': 0,
    'max': 3000, # Ajusta según el brillo que necesites
    'gamma': 1.4
}
```

```python
raw_coord = [-69.6728045225969, -0.566948200364969]
loc_agua = ee.Geometry.Point(raw_coord)

bandas = ['B3', 'B4', 'B8', 'B11']
Valores = imagen_limpia.select(bandas).reduceRegion(
    reducer=ee.Reducer.mean(),
    geometry=loc_agua,
    scale=10
).getInfo()

for banda, valor in Valores.items():
    print(f"Reflectancia real en {banda}: {valor * 0.0001}")

```
